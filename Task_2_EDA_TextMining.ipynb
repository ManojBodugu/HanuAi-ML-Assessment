```python
import pandas as pd
from pathlib import Path

# Load the assessment dataset
file_path = Path("Hanuai_task2_assignment.csv")
df = pd.read_csv(file_path)

# Check the dataset
print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

display(df.head())
```

```python
# HANUAI ML ASSESSMENT - TASK 2
# Step 2: Data quality assessment

# 1. Check column names, data types, and non-null counts
print("=== DATASET INFORMATION ===")
df.info()

# 2. Count missing values in every column
print("\n=== MISSING VALUES ===")
missing = df.isnull().sum()

missing_report = pd.DataFrame({
    "Missing_Count": missing,
    "Missing_Percentage": (missing / len(df) * 100).round(2)
})

display(missing_report.sort_values(
    "Missing_Count", ascending=False
))

# 3. Check duplicate rows
print("\n=== DUPLICATE CHECK ===")
print("Duplicate complete rows:", df.duplicated().sum())

# 4. Check duplicate event IDs
print("Duplicate Event IDs:", df["Event id"].duplicated().sum())

# 5. Check date values
print("\n=== DATE CHECK ===")
dates = pd.to_datetime(df["Opened date"], errors="coerce")
print("Valid dates:", dates.notna().sum())
print("Invalid or missing dates:", dates.isna().sum())
print("Earliest date:", dates.min())
print("Latest date:", dates.max())
```

```python
# HANUAI ML ASSESSMENT - TASK 2
# Step 3: Explore complaint patterns and event trends

# Convert event dates to datetime
df["Opened date"] = pd.to_datetime(
    df["Opened date"], errors="coerce"
)

# Display the most common complaint descriptions
print("TOP 10 COMPLAINT DESCRIPTIONS")

complaint_counts = (
    df["COMPLAINT_CD_DESC"]
    .fillna("Unknown")
    .value_counts()
    .head(10)
)

display(complaint_counts)

# Chart 1: Most frequent complaint descriptions
plt.figure(figsize=(12, 6))
complaint_counts.sort_values().plot(kind="barh")
plt.title("Top 10 Complaint Descriptions")
plt.xlabel("Number of Events")
plt.ylabel("Complaint Description")
plt.tight_layout()
plt.show()

# Chart 2: Events recorded per month
monthly_events = (
    df.dropna(subset=["Opened date"])
      .set_index("Opened date")
      .resample("MS")
      .size()
)

plt.figure(figsize=(12, 5))
monthly_events.plot(kind="line", marker="o")
plt.title("Monthly Event Frequency")
plt.xlabel("Month")
plt.ylabel("Number of Events")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# HANUAI ML ASSESSMENT - TASK 2
# Step 4: Text mining and issue classification

import re

# Combine narrative fields for text mining
text_columns = [
    "CAUSAL_VERBATIM",
    "CORRECTION_VERBATIM",
    "CUSTOMER_VERBATIM",
    "COMPLAINT_CD_DESC",
    "CAUSAL_CD_DESC"
]

# Keep only columns that exist in the dataset
text_columns = [c for c in text_columns if c in df.columns]

df["combined_text"] = (
    df[text_columns]
    .fillna("")
    .astype(str)
    .agg(" ".join, axis=1)
    .str.lower()
)

# Preliminary rule-based issue classification
def classify_issue(text):
    text = str(text).lower()

    if any(word in text for word in [
        "wire", "wiring", "voltage", "battery", "electrical", "power"
    ]):
        return "Electrical Issue"

    if any(word in text for word in [
        "software", "update", "programming", "firmware", "reset"
    ]):
        return "Software / Programming"

    if any(word in text for word in [
        "communication", "signal", "connection", "network", "radio"
    ]):
        return "Communication / Connectivity"

    if any(word in text for word in [
        "screen", "display", "video", "black screen"
    ]):
        return "Display / Video"

    if any(word in text for word in [
        "noise", "rattle", "vibration", "sound"
    ]):
        return "Noise / Vibration"

    if any(word in text for word in [
        "broken", "crack", "failed", "failure", "damaged"
    ]):
        return "Component / Hardware Failure"

    return "Other / Needs Review"

df["issue_category"] = df["combined_text"].apply(classify_issue)

# Extract simple keyword tags from the narrative
keywords = [
    "radio", "screen", "display", "audio", "video",
    "communication", "battery", "wiring", "software",
    "noise", "failure", "connection"
]

def extract_tags(text):
    return [
        keyword for keyword in keywords
        if re.search(r"\b" + re.escape(keyword) + r"\b", text)
    ]

df["extracted_tags"] = df["combined_text"].apply(extract_tags)

# Review the results
print("Issue category counts:")
display(df["issue_category"].value_counts().to_frame("Event_Count"))

display(df[[
    "Event id", "COMPLAINT_CD_DESC",
    "issue_category", "extracted_tags"
]].head(10))
```

```python
# Export enriched dataset for submission
output_path = Path("Hanuai_Task2_Enriched.csv")
df.to_csv(output_path, index=False)

print("Saved enriched dataset to:", output_path.resolve())
```

```python
# Step 5: Cluster similar service complaints using TF-IDF and K-Means

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans

# Convert narrative text into numerical TF-IDF features
texts = df["combined_text"].fillna("").astype(str)

vectorizer = TfidfVectorizer(
    stop_words="english",
    max_features=1000,
    ngram_range=(1, 2),
    min_df=2
)

X = vectorizer.fit_transform(texts)

# Group complaints into 5 exploratory clusters
kmeans = KMeans(n_clusters=5, random_state=42, n_init=10)
df["complaint_cluster"] = kmeans.fit_predict(X)

# Show the most important terms in each cluster
terms = vectorizer.get_feature_names_out()

for cluster_id, center in enumerate(kmeans.cluster_centers_):
    top_indices = center.argsort()[-10:][::-1]
    top_terms = [terms[i] for i in top_indices]
    print(f"Cluster {cluster_id}: {', '.join(top_terms)}")

print("\nRecords per cluster:")
display(df["complaint_cluster"].value_counts().sort_index().to_frame("Record_Count"))
```

```python
# Task 2: Cluster similar complaints using TF-IDF and K-Means

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans

# Convert complaint narratives into numerical features
texts = df["combined_text"].fillna("").astype(str)

vectorizer = TfidfVectorizer(
    stop_words="english",
    max_features=1000,
    ngram_range=(1, 2),
    min_df=2
)

X = vectorizer.fit_transform(texts)

# Create five exploratory complaint clusters
model = KMeans(n_clusters=5, random_state=42, n_init=10)
df["complaint_cluster"] = model.fit_predict(X)

# Print representative terms for each cluster
terms = vectorizer.get_feature_names_out()

for cluster_id, center in enumerate(model.cluster_centers_):
    top_indices = center.argsort()[-10:][::-1]
    print(
        f"Cluster {cluster_id}: "
        + ", ".join(terms[i] for i in top_indices)
    )

print("\nRecords per cluster:")
display(df["complaint_cluster"].value_counts().sort_index())
```
